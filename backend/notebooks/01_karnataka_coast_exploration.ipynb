## 2. Major Fishing Ports & Landing Centers Along Karnataka

## 3. Distance Matrix Between Selected Landing Centers

```python
malpe = KARNATAKA_LANDING_CENTERS["Malpe"]
mangalore = KARNATAKA_LANDING_CENTERS["Mangalore"]
karwar = KARNATAKA_LANDING_CENTERS["Karwar"]

print(f"Distance Malpe -> Mangalore: {haversine_distance_km(malpe, mangalore)} km")
print(f"Distance Malpe -> Karwar:    {haversine_distance_km(malpe, karwar)} km")
```
